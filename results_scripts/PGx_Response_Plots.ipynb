# Figures for PGx LLM Eval

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Figures/table3_agreement_scores.csv')
data
sns.set_theme(style='white', palette='deep')
ax = sns.barplot(data = data, x = 'Scorer', y = 'Agreement F1')
ax.set_title('Concordance of Response Similarity - F1 Scores')
ax.tick_params(axis='x', labelrotation=90)
ax.bar_label(ax.containers[0], fontsize=8)
plt.savefig('Fig1_agreementF1.svg', bbox_inches = "tight")
```

```python
dat = pd.read_csv('Figures/table4_miss_wellspecified_refusal.csv')
dat = dat.rename(columns = {'Yes': 'Misspecified', 'No': 'Not Misspecified'})
dat = dat.round(3)
ax = dat.plot(x="Model", y=["Misspecified", "Not Misspecified"], kind="bar", xlabel = 'Model', ylabel = 'Refusal Rate', rot=0)
ax.set_title('Average Refusal Rate for Misspecified and Well Specified Queries')
ax.bar_label(ax.containers[0], fontsize=6)
ax.bar_label(ax.containers[1], fontsize=6)
plt.savefig('Fig2_refusalrate.svg', bbox_inches = "tight")
```
